## 1. Trabajando con NumPy

```python
#Import de la libreria.

import numpy as np

#Array del 10 al 100 incluyendo estos dos.

array = np.arange(10,101)
print(array)

#Array del 10 al 100 al reves.

print(np.flip(array))

#Array con valores del 0 al 8 en 2d (3x3).

array_2d_3x3 = np.arange(9).reshape(3, 3)
print(array_2d_3x3)

#Array identidad en 2d (6x6).

array_identidad = np.eye(6)
print(array_identidad)

#Array con valores random en 3d (3x3x3).

array_3d_random = np.random.random((3,3,3))
print(array_3d_random)

#Array randoms y comparacion.

arrayA = np.random.random(5)
arrayB = np.random.random(5)
comparacion = np.array_equal(arrayA, arrayB)
if comparacion:
    print("Los Arrays son iguales")
else:
    print("Los Arrays no son iguales")
```

## 2. NumPy: conversión de valores

```python
#Import de la libreria.

import numpy as np

#Array de dolares del 1 al 1000 con 100 valores.

dolares = np.random.randint(1, 1001, 100)

#Conversion de dolares a euros y redondeo.

euros = np.round(dolares * 0.874413, 2)

#Prints de los dos arrays.

print(dolares)
print(euros)
```

## 3. pandas y dataset tips de seaborn

```python
#Import de todas las librerias usadas.

import pandas as pd
import seaborn as sns

#Declaracion de la variable df.

df = sns.load_dataset("tips")

#Muestra los 5 primeros registros.

display(df.head(5))

#Descripción de los atributos del dataset:

print("total_bill: importe total de la cuenta (USD).\ntip: propina (USD).\nsex: sexo de la persona que pagó.\nsmoker: si había fumadores en la mesa.\nday: día de la semana.\ntime: momento del día.\nsize: número de personas en la mesa.")

#Muestra el número total de registros.

print(len(df))

#Calcula la propina media y la cuenta media.

print(f"La media de la propina es de: {df["tip"].median().round(2)} y la media de la cuenta es de: {df["total_bill"].median().round(2)}")

#Crea la columna pct_tip y muestra las 5 filas con mayor porcentaje.

df["pct_tip"] = ((df["tip"] / df["total_bill"]) * 100).round(2)
print(df.sort_values("pct_tip", ascending = False).head(5))

#Muestra las 10 cuentas más altas con su propina y porcentaje.

print(df.sort_values("total_bill", ascending = False).head(10)[["total_bill", "tip", "pct_tip"]])

#Muestra el promedio de propina por día ordenado de mayor a menor.

print(df.groupby("day")["tip"].mean().round(2))

#Muestra el promedio de pct_tip por sexo y por condición de fumador.

print(df.groupby(["smoker", "sex"])["pct_tip"].mean().round(2))

#Indica el día y momento en el que se registró la cuenta más alta.

print(df.loc[df["total_bill"].idxmax()][["total_bill", "day", "time"]])

#Muestra cuántos registros corresponden a Lunch y Dinner.

print(df["time"].value_counts())

#Muestra la propina media por tamaño de mesa e indica cuál deja mayor porcentaje.

print(df.groupby("size")["pct_tip"].mean().round(2))

#Filtra mesas de 5 o más personas y calcula su pct_tip medio.

print(df[df["size"] >= 5]["pct_tip"].mean().round(2))

#Calcula, por día, la facturación total y el total de propinas.

print(df.groupby("day")[["total_bill", "tip"]].sum().round(2))

#Calcula la correlación entre total_bill y tip y comenta el resultado.

print(df["total_bill"].corr(df["tip"]).round(2))

#Exporta a CSV las columnas indicadas en un archivo tips_resumen.csv.

df[["total_bill", "tip", "pct_tip", "day", "time", "size"]].to_csv("tips_resumen.csv", index=False)
```
